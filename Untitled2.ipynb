```python
years = [str(year) for year in range(2008, 2017)]

for year in years:
    row = df.loc[df[year].idxmax()]
    print(year, ":", row['States/UTs'], "-", row[year])
```

```python
for year in years:
    row = df.loc[df[year].idxmin()]
    print(year, ":", row['States/UTs'], "-", row[year])
```

```python
year_totals = df[years].sum()

year_totals
```

```python
print("Total values by year:")
print(year_totals)
```

```python
plt.figure(figsize=(10,5))

plt.plot(year_totals.index, year_totals.values, marker='o')

plt.title("Year-wise Trend of Persons Injured Due to Exceeding Lawful Speed")
plt.xlabel("Year")
plt.ylabel("Number of Persons Injured")

plt.grid(True)
plt.show()
```

```python
df['Total_2008_2016'] = df[years].sum(axis=1)
```

```python
df[['States/UTs', 'Total_2008_2016']]
```

```python
top_states = df.sort_values(
    'Total_2008_2016',
    ascending=False
)

top_states[['States/UTs', 'Total_2008_2016']].head(10)
```

```python
top_states = df[df['States/UTs'] != 'All India'].sort_values(
    'Total_2008_2016',
    ascending=False
)

top10 = top_states.head(10)
```

```python
top10[['States/UTs', 'Total_2008_2016']]
```

```python
comparison = df[['States/UTs', '2008', '2016']].copy()

comparison['Change'] = comparison['2016'] - comparison['2008']

comparison.sort_values('Change', ascending=False).head(10)
```

```python
comparison.sort_values('Change').head(10)
```

```python
top10_comparison = df.loc[
    df['States/UTs'].isin(top10['States/UTs']),
    ['States/UTs', '2008', '2016']
]

top10_comparison.set_index('States/UTs').plot(
    kind='bar',
    figsize=(12,6)
)

plt.title("Persons Injured: 2008 vs 2016")
plt.xlabel("State/UT")
plt.ylabel("Persons Injured")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
